```python
def vacuum_cleaner_agent_simulation(initial_state):
    current_room, room_A_status, room_B_status = initial_state
    print(f"state0 - ({current_room},{room_A_status},{room_B_status})")

    actions_taken = []

    while room_A_status == 'D' or room_B_status == 'D':
        if current_room == 'A':
            if room_A_status == 'D':
                actions_taken.append("Clean A")
                print("action - clean dust (A)")
                room_A_status = 'C'

            if room_B_status == 'D':
                actions_taken.append("Move B")
                print("action - move to B")
                current_room = 'B'
            else:
                break

        elif current_room == 'B':
            if room_B_status == 'D':
                actions_taken.append("Clean B")
                print("action - clean dust (B)")
                room_B_status = 'C'

            if room_A_status == 'D':
                actions_taken.append("Move A")
                print("action - move to A")
                current_room = 'A'
            else:
                break

        print(f"state - ({current_room},{room_A_status},{room_B_status})")

    actions_taken.append("Done.")
    print("Done.")
    return actions_taken

print("\n--- Scenario 1: Room A:Dirty, Room B:Dirty, start A ---")
scenario1_actions = vacuum_cleaner_agent_simulation(('A', 'D', 'D'))
print("Log:", scenario1_actions)

print("\n--- Scenario 2: Room A:Dirty, Room B:Dirty, start B ---")
scenario2_actions = vacuum_cleaner_agent_simulation(('B', 'D', 'D'))
print("Log:", scenario2_actions)

```
