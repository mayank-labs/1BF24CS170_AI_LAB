```python
import math


def alpha_beta_search(tree, maximizing_player=True):
  pruned_paths = []

  def max_value(node, alpha, beta, path):
    if isinstance(node, (int, float)):
      return node, path

    v = -math.inf
    best_path = []

    for i, child in enumerate(node):
      child_path = f"{path} -> Child {i+1}"
      val, current_path = min_value(child, alpha, beta, child_path)

      if val > v:
        v = val
        best_path = current_path

      if v >= beta:
        for rem in range(i + 1, len(node)):
          pruned_paths.append(
              f"{path} -> Child {rem+1} (and its subtree) [Beta cutoff: v={v} >="
              f" beta={beta}]"
          )
        return v, best_path

      alpha = max(alpha, v)

    return v, best_path

  def min_value(node, alpha, beta, path):
    if isinstance(node, (int, float)):
      return node, path

    v = math.inf
    best_path = []

    for i, child in enumerate(node):
      child_path = f"{path} -> Child {i+1}"
      val, current_path = max_value(child, alpha, beta, child_path)

      if val < v:
        v = val
        best_path = current_path

      if v <= alpha:
        for rem in range(i + 1, len(node)):
          pruned_paths.append(
              f"{path} -> Child {rem+1} (and its subtree) [Alpha cutoff: v={v}"
              f" <= alpha={alpha}]"
          )
        return v, best_path

      beta = min(beta, v)

    return v, best_path

  if maximizing_player:
    root_val, optimal_path = max_value(tree, -math.inf, math.inf, "Root")
  else:
    root_val, optimal_path = min_value(tree, -math.inf, math.inf, "Root")

  return root_val, optimal_path, pruned_paths


game_tree = [
    [[[5, 6], [7, 4, 5]], [[3]]],
    [[[6], [6, 9]], [[7]]],
    [[[5]], [[9, 8], [6]]],
]

value, path, pruned = alpha_beta_search(game_tree, maximizing_player=True)

print(f"Root Node Value: {value}")
print(f"Optimal Path:    {path}")
print("\nPruned Paths:")
for p in pruned:
  print(f" - {p}")
```
