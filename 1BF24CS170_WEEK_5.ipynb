```python
import math
import random

def objective_function(x):
    return x ** 2

def simulated_annealing_with_tracking(objective, bounds, n_iterations, init_temp, cooling_rate):
    current_sol = random.uniform(bounds[0], bounds[1])
    current_eval = objective(current_sol)
    best_sol = current_sol
    best_eval = current_eval
    temp = init_temp

    print(f"{'Iteration':<10} | {'Temperature':<12} | {'Current Eval':<18}")
    print("-" * 50)

    for i in range(n_iterations):
        step_size = (bounds[1] - bounds[0]) * 0.1 * (temp / init_temp)
        candidate_sol = current_sol + random.uniform(-step_size, step_size)
        candidate_sol = max(bounds[0], min(bounds[1], candidate_sol))
        candidate_eval = objective(candidate_sol)

        # Acceptance logic
        if candidate_eval < current_eval:
            current_sol = candidate_sol
            current_eval = candidate_eval
            if candidate_eval < best_eval:
                best_sol = candidate_sol
                best_eval = candidate_eval
        else:
            diff = candidate_eval - current_eval
            exponent = -diff / max(temp, 1e-10)
            # Prevent float overflow
            exponent = max(min(exponent, 700), -700)
            acceptance_prob = math.exp(exponent)
            if random.random() < acceptance_prob:
                current_sol = candidate_sol
                current_eval = candidate_eval

        # Print temperature drop every 10 iterations (and first 3 steps)
        if i % 10 == 0 or i < 3:
            print(f"{i:<10} | {temp:<12.4f} | {current_eval:<18.6f}")

        # Geometric Cooling Schedule: decrease temperature
        temp *= cooling_rate

    print("-" * 50)
    return best_sol, best_eval

# Run tracking demo
if __name__ == "__main__":
    best_sol, best_eval = simulated_annealing_with_tracking(
        objective=objective_function,
        bounds=[-10.0, 10.0],
        n_iterations=100,
        init_temp=100.0,
        cooling_rate=0.95
    )
    print(f"Best Solution: {best_sol:.6f}, Best Evaluation: {best_eval:.6f}")
```
